## One-to-Many

```python
folder = 'one_to_many'
```

```python
#Create GT
gt = np.zeros((500, 500), dtype=np.int32)

# Draw circle (GT object 1)
rr, cc = disk((250, 150), 120)
gt[rr, cc] = 1

# Draw circle (GT object 2)
rr, cc = disk((250, 370), 100)
gt[rr, cc] = 2

#Save
np.save(f'../mma/datasets/synthetic test cases/{folder}/gt.npy', gt)

#Plot
plt.imshow(gt)
plt.show()
```

```python
# Pred + GT (for visualization)
fig, ax = plt.subplots(figsize=(6,6))

# Background
ax.set_facecolor("white")



#------------------
#   Pred Disks
#------------------
pred_mask = np.zeros((500, 500), dtype=bool)

pred_circles = [
    ((270, 250), 80)
]

for center, radius in pred_circles:
    disk_center = (center[1], center[0])
    rr, cc = disk(disk_center, radius)
    pred_mask[rr, cc] = True

    circle = Circle(
        center,
        radius,
        facecolor="#ADD8E6",
        # facecolor="#DF6060",
        # edgecolor="none",
        edgecolor="#56C2E6",
        linewidth=2.5,
        alpha=0.6
    )
    ax.add_patch(circle)

#------------------
#   GT Circles
#------------------
gt_mask = np.zeros((500, 500), dtype=bool)

gt_circles = [
   ((150, 250), 120),
   ((370, 250), 100)
]

for center, radius in gt_circles:
    disk_center = (center[1], center[0])
    rr, cc = disk(disk_center, radius)
    gt_mask[rr, cc] = True

    circle = Circle(
        center,
        radius,
        # facecolor="#cccccc",
        facecolor="none",
        edgecolor="black",
        linewidth=2.5,
        linestyle="--",   # dashed
        alpha=0.6
    )
    ax.add_patch(circle)



#------------------
#    Visualize
#------------------
ax.set_xlim(0, 500)
ax.set_ylim(500, 0)
ax.set_aspect('equal')
plt.show()
```

```python
#Create Pred
pred = np.zeros((500, 500), dtype=np.int32)

#Draw circle (Pred Object)
rr, cc = disk((250, 270), 60)
pred[rr, cc] = 1

#Save Pred
np.save(f'../mma/datasets/synthetic test cases/{folder}/pred.npy', pred)

plt.imshow(pred)
plt.show()
```

# Collect Metrics on Synthetic Examples

## Utils

```python
from scipy.ndimage import binary_dilation
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

from mma.corruption_functions.mask_erosion import get_border_coords

def create_accuracy_plot(accuracy_plot, pred, gt, title, border_thickness=1):
    accuracy_plot[0][0] = 2
    accuracy_plot[0][1] = 1

    colors = ['white', 'red', 'green', 'black', 'black']
    cmap = mcolors.ListedColormap(colors)

    # -------------------
    # GT Borders
    # -------------------
    for gt_label in np.unique(gt):
        if gt_label != 0:

            border_mask = np.zeros_like(gt, dtype=bool)

            row_coords, col_coords = get_border_coords(
                np.where(gt == gt_label, 1, 0)
            )

            border_mask[row_coords, col_coords] = True

            # Thicken border
            thick_border = binary_dilation(
                border_mask,
                iterations=border_thickness
            )

            accuracy_plot[thick_border] = 3

    # -------------------
    # Pred Borders
    # -------------------
    for pred_label in np.unique(pred):
        if pred_label != 0:

            border_mask = np.zeros_like(pred, dtype=bool)

            row_coords, col_coords = get_border_coords(
                np.where(pred == pred_label, 1, 0)
            )

            border_mask[row_coords, col_coords] = True

            thick_border = binary_dilation(
                border_mask,
                iterations=border_thickness
            )

            accuracy_plot[thick_border] = 4

    plt.imshow(accuracy_plot, cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
def collect_metrics(example_name):
    gt = np.load(f'../mma/datasets/synthetic test cases/{example_name}/gt.npy')
    pred = np.load(f'../mma/datasets/synthetic test cases/{example_name}/pred.npy')

    mma, mma_plot = get_mma(pred, gt, return_accuracy_plot=True)
    mma_greedy, mma_greedy_plot = get_mma(pred, gt, greedy=True, return_accuracy_plot=True)
    seg, seg_plot = get_seg(pred, gt, return_accuracy_plot=True)
    pq, pq_plot = get_pq(pred, gt, return_accuracy_plot=True)
    ap_50, ap_50_plot = get_ap_50(pred, gt, return_accuracy_plot=True)
    aji, aji_plot = get_aji(pred, gt, return_accuracy_plot=True)

    print(f'MMA: {mma}')
    print(f'MMA Greedy: {mma_greedy}')
    print(f'SEG: {seg}')
    print(f'PQ: {pq}')
    print(f'AP@50: {ap_50}')
    print(f'AJI: {aji}')

    create_accuracy_plot(mma_plot, pred, gt, 'MMA')
    create_accuracy_plot(mma_greedy_plot, pred, gt, 'MMA Greedy')
    create_accuracy_plot(seg_plot, pred, gt, 'SEG')
    create_accuracy_plot(pq_plot, pred, gt, 'PQ')
    create_accuracy_plot(ap_50_plot, pred, gt, 'AP@50')
    create_accuracy_plot(aji_plot, pred, gt, 'AJI')
```
