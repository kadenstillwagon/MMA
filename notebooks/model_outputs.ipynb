```python
import numpy as np
from tqdm import tqdm
from PIL import Image
import os
from mma.segmentation_models.dinocell import get_dinocell_output
from mma.segmentation_models.cellpose_sam import get_cellpose_sam_output
```

```python
#Load LIVECell Images and GT

imgs = np.load('../mma/datasets/LiveCell/test/imgs.npy')
anns = np.load('../mma/datasets/LiveCell/test/anns.npy')
```

```python
#DINOCell Outputs
for i in range(len(imgs)):
    img = Image.fromarray(imgs[i])
    img_path = f'../mma/dataset/LiveCell/test/img_{i}.png'
    img.save(img_path)
    dinocell_output = get_dinocell_output(img_path)
    np.save(f'../mma/livecell model outputs/DINOCell/output_{i}.npy', np.array(dinocell_output))
    os.remove(img_path)
```

```python
#Cellpose-SAM Outputs
cp_sam_preds = get_cellpose_sam_output(imgs)
for i in tqdm(range(len(cp_sam_preds))):
    np.save(f'../mma/livecell model outputs/Cellpose-SAM/output_{i}.npy', np.array(cp_sam_preds[i]))
```
